# Autoregressive Bigram 训练

使用同一 tokenizer 和预处理数据拟合 Bigram 基线，并保存独立推理 artifact。

## 配置和数据

```python
from pathlib import Path

from scripts.experiment import (
    AutoregressiveBigramConfig,
    DataConfig,
    DeviceConfig,
    ExperimentConfig,
)
from scripts.pipeline import (
    build_dataloaders,
    build_model,
    load_tokenizer,
    save_inference_artifact,
    save_training_artifact,
)

config = ExperimentConfig(
    data=DataConfig(
        dataset_path=Path("data/processed"),
        tokenizer_path=Path("data/processed/tokenizer.json"),
        output_dir=Path("output/autorg_bigram"),
    ),
    model=AutoregressiveBigramConfig(alpha=1.0),
    device=DeviceConfig(mode="formal_cuda"),
)
tokenizer = load_tokenizer(config)
train_loader, _, _ = build_dataloaders(config, tokenizer)
```

## 拟合和保存

```python
model = build_model(config, tokenizer)
model.fit(train_loader.dataset, verbose=True, batch_size=4096)
```

```python
save_training_artifact(config, tokenizer, {"count_sum": int(model.count.sum())})
save_inference_artifact(config, model, tokenizer)
print(f"Saved Bigram artifact to {config.data.output_dir}")
```
